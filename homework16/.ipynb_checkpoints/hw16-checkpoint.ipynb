```python
import data
import matplotlib.pyplot as plt
import torch 

model = torch.nn.Sequential(
    torch.nn.Linear(2,50), 
    torch.nn.Sigmoid(), 
    torch.nn.Linear(50,1) 
)

loss_fn = torch.nn.MSELoss()

optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

for t in range(1000):
    y_pred = model(data.x)
    loss = loss_fn(y_pred, data.y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if t % 100 == 0:
        print(f"Iteration {t}, Loss: {loss.item()}")
```

What is the difference between artificial intelligence, machine learning and deep learning?

```python
# Artificial Intelligence is creating smart, intelligent machines. Machine learning is a type of artificial intelligence
# That helps make AI-run-Applications. Deep learning is a type of Machine Learning that uses data & complicated algorithms
# To train a "model".
```

What is a neural network?

```python
# A neural network is a type of model in machine learning that tries to mimic the brain's algorithm. It has nodes called
# 'Artificial neurons'.
```

In lecture, we used `torch.nn.Linear(2, 50)`. What do the numbers `2` and `50` represent? 

```python
# The 2 represents the Input layer, the 50 represents the hidden layer
# Model takes 2 inputs and passes it through 50 neurons.
```

What does the activation function do?

```python
# It introduces non-linearity to the model.
```

What is a loss function? Why do we need one?

```python
# The loss function is a measure of how wrong the model is. We need one to know how accurate & whether we should optimize the model or not.
```

Why do we call `optimizer.zero_grad()` before calling `loss.backward()`?

```python
# opitmizer.zero_grad resets the gradients, and loss.backward() computes the new gradient
# This is like resetting your variables.
```

```python
model = torch.nn.Sequential(
    torch.nn.Linear(2,50), 
    torch.nn.ReLU(), 
    torch.nn.Linear(50,1) 
)

loss_fn = torch.nn.MSELoss()

optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

for t in range(1000):
    y_pred = model(data.x)
    loss = loss_fn(y_pred, data.y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if t % 100 == 0:
        print(f"Iteration {t}, Loss: {loss.item()}")
# Loss function values slightly less than original
```

Change your activation function from `torch.nn.Sigmoid()` to `torch.nn.Tanh`? Compare the loss functions from lecture to your modification on the **1000th** iteration. 

```python
model = torch.nn.Sequential(
    torch.nn.Linear(2,50), 
    torch.nn.Tanh(), 
    torch.nn.Linear(50,1) 
)

loss_fn = torch.nn.MSELoss()

optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

for t in range(1000):
    y_pred = model(data.x)
    loss = loss_fn(y_pred, data.y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if t % 100 == 0:
        print(f"Iteration {t}, Loss: {loss.item()}")
# Loss function significantly less than original.
```

What is the difference between `torch.nn.Sigmoid()`, `torch.nn.ReLU` and `torch.nn.Tanh`? 

```python
# Sigmoid produces the least accurate result, ReLU produces more accurate results, and Tanh produces the most accurate result.
```

# 3) More Advanced
Create a new file called `new_data.py`, make a new "hidden" function and add noise/ obscure the data (e.g. like what is done in `data.py` from lecture). Import that file here.

```python
import new_data
```

Define a new model and train it **poorly** in the cell below.

```python
# Poorly trained: 25 neurons, learning rate: 1e-4, Sigmoid activation function, 500 iterations.
model_poor = torch.nn.Sequential(
    torch.nn.Linear(2,25), 
    torch.nn.Sigmoid(), 
    torch.nn.Linear(25,1) 
)

loss_fn = torch.nn.MSELoss()

optimizer = torch.optim.Adam(model_poor.parameters(), lr=1e-4)

for t in range(500):
    y_pred = model_poor(new_data.x)
    loss = loss_fn(y_pred, new_data.y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if t % 100 == 0:
        print(f"Iteration {t}, Loss: {loss.item()}")
```

Improve your model and train it **well** in the cell below. (You can add more neurons, iterations, etc).

```python
# Well trained: 75 neurons, Tanh activation function, learning rate 1e-2, 2000 iterations
model_well = torch.nn.Sequential(
    torch.nn.Linear(2,75), 
    torch.nn.Tanh(), 
    torch.nn.Linear(75,1) 
)

loss_fn = torch.nn.MSELoss()

optimizer = torch.optim.Adam(model_well.parameters(), lr=1e-2)

for t in range(2000):
    y_pred = model_well(new_data.x)
    loss = loss_fn(y_pred, new_data.y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if t % 100 == 0:
        print(f"Iteration {t}, Loss: {loss.item()}")
```

Make foure subplots (two rows, two columns). 
- (TOP LEFT) Show what the "hidden pattern" is supposed to be.
- (TOP RIGHT) Show that your "data" is.
- (BOTTOM LEFT) Show your model's prediction when you trained it poorly.
- (BOTTOM RIGHT) Show your model's prediction when you trained it well.

```python
fig, ((ax1,ax2),(ax3,ax4)) = plt.subplots(2,2)

new_x = torch.linspace(-10, 10, 100) 
grid = torch.stack(torch.meshgrid((new_x, new_x)), dim=2) 

ax1.imshow(new_data.pattern)

ax2.scatter(new_data.x_values, new_data.y_values, c=new_data.color_values)

output3 = model_poor(grid).detach() 
output3 = output3.squeeze(2).t()
ax3.imshow(output3, extent=(-10,10,-10,10))

output4 = model_well(grid).detach() 
output4 = output4.squeeze(2).t() 

ax3.imshow(output3, extent=(-10,10,-10,10))
ax4.imshow(output4, extent=(-10,10,-10,10))
plt.show()      
```
