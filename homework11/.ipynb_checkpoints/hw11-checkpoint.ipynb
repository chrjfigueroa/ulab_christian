## 1 Functions
**Orbital Periods and Kepler's Third Law** \
Kepler's Third Law states that the square of the orbital period $T$ (in years) of a planet is proportional to the cube of the semi-major axis $a$ (in astronomical units, AU):

$$
T^2 = a^3
$$

Write a **function** that calculates the orbital period $T$ given the semi-major axis $a$. (So, you can assume that $T = \sqrt{a^3}$.)

```python
#Parameter semaj is the semimajor axis length in AU
#Functions returns orbital period T in Years.
def period(semmaj):
    T = semmaj**(3/2)
    return T
```

Call your function with the following values: 
* `planet_a` $ = 0.5\,\text{AU}$
* `planet_b` $ = 1.0\,\text{AU}$
* `planet_c` $ = 1.5\,\text{AU}$

```python
# length of planets a,b,c semimajor axes.
planet_a = 0.5
planet_b = 1.0
planet_c = 1.5

print(period(planet_a))
print(period(planet_b))
print(period(planet_c))
#returns the period in Years of planet a b and c respectively.
```

Rewrite your **function** from above with a `numpy` function. 

```python
#Takes semaj as a parameter equivalent to planet's semmimajor axis length in AU
#Returns T as the orbital period of the Planet in Years.
def orb_period(semmaj):
    T = np.power(semmaj,3/2)
    return T
```

## 2 Indexing + Conditionals

You are provided with a list of exoplanet radii (in Earth radii) and their equilibrium temperatures (in Kelvin):

`radii = [1.2, 0.8, 1.5, 2.0, 1.0, 0.6]` \
`temperatures = [300, 400, 350, 280, 500, 450]`

Write a loop to filter and print the radii of planets with temperatures below 350 K.

```python
radii = [1.2, 0.8, 1.5, 2.0, 1.0, 0.6]
temperatures = [300, 400, 350, 280, 500, 450]

small_cold_radii = [] #empty list that will contain the radii of the small and cold planets


for x,temp in enumerate(temperatures): 
    #sorts through the temperatures & appends radii of exoplanets who are less than 350k to small_cold_radii.
    if temp<350:
        small_cold_radii.append(radii[x])

print(small_cold_radii)
        
```

Use list slicing to extract and print the radii of the first three planets.

```python
print(radii[0:3])
```

Turn the radii and temperatures data into a dictionary. 

```python
planet_data = {}
for i in range(6):
    planet_data.update({f'{i+1}': [radii[i],temperatures[i]]})
print(planet_data)
#gives each planet a number key and returns a list per key that corresponds to the radius and temp of planet.
```

Give each of the planets in your dictionary a name. 

```python
names = ['Jupyter4','Earth2','Mars10','Mercury_5','PlutoBig','NeptuneGod']
for i in range(6):
    planet_data[f"{names[i]}"] = planet_data.pop(f"{i+1}")
print(planet_data['Jupyter4'])
print(planet_data['Earth2'])
print(planet_data['Mars10'])
print(planet_data['Mercury_5'])
print(planet_data['PlutoBig'])
print(planet_data['NeptuneGod'])
#renames keys to the planet names
```

Create a new dictionary of planets with radii between 1.0 and 2.0 Earth radii (inclusive).

```python
small_radii_planets = {}
for i,key in enumerate(planet_data):
    
    if (1<=planet_data[f'{key}'][0]<=2): #if radius of planets in planet_data dictionary is between 1 and 2 earth radii (inclusive)
        small_radii_planets.update({f'Planet {i+1}': planet_data[f'{key}']}) # add this planet data to the small_radii planets dictionary.
        
print(small_radii_planets)
```

## 4 Plotting
Create a scatter plot of the following semi-major axes (x-axis) vs. orbital periods (y-axis). Label the axes, add a legend and add a title: "Orbital Periods of Exoplanets."

`semi_major_axes = [0.5, 1.0, 2.0, 1.5, 1.2, 0.8]`

```python
import matplotlib.pyplot as plt
semi_major_axes = [0.5, 1.0, 2.0, 1.5, 1.2, 0.8] # list
semi_maj = np.array(semi_major_axes) #np array
def orb_period(semi_maj):
    T = np.power(semi_maj,3/2)
    return T
orbital_period = orb_period(semi_maj) #np array

plt.figure()
plt.scatter(semi_maj,orbital_period, label = r'$T^{2} = a^{3}$')
plt.xlabel('Semi-major Axis')
plt.ylabel('Orbital Period')
plt.title('Orbital Periods of Exoplanets')
plt.legend()
plt.show()
```

Create a bar plot of planet radii vs. their equilibrium temperatures. Label the axes, add a legend and add a title: "Equilibrium Temperatures vs. Radii."

```python
plt.figure()
plt.bar(radii,temperatures, label="Planet Radii and Equilibrium Temps")
plt.xlabel('Radius')
plt.ylabel('Equilibrium Temperatures')
plt.title('Equilibrium Temperatures vs. Radii')
plt.legend()
plt.show()
```

Put the scatter plot and bar plot on the same figure (horizonatally). 

```python
fig, (scatt, bar) = plt.subplots(1,2)
scatt.scatter(semi_maj,orbital_period, label = r'$T^{2} = a^{3}$')
scatt.set_xlabel('Semimajor Axis')
scatt.set_ylabel('Orbital Period')
scatt.set_title('Orbital Periods of Exoplanets')
scatt.legend()

bar.bar(radii,temperatures, label="Planet Radii and Equilibrium Temps")
bar.set_xlabel('Radius')
bar.set_ylabel('Equilibrium Temperatures')
bar.set_title('Equilibrium Temperatures vs. Radii')
bar.legend(loc='lower left')

fig.tight_layout(pad=1)
plt.show()
```
